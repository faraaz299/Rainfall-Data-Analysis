```python
#Finding the maximum and minimum temperature
max = df['maxtemp'].max()
print(max)
min = df['maxtemp'].min()
print(min)
```

```python
#finding when these temperatures occurred
df.loc[df['maxtemp'].idxmax()]
```

```python
df.loc[df['maxtemp'].idxmin()]
```

```python
#finding 10 hottest days
hottest_10 = df.sort_values('maxtemp', ascending=False).head(10)
hottest_10
```

```python
#calculating how many of the 10 hottest days were rainy.
hottest_10['rainfall'].value_counts()
```

```python
##Calculating the percentage of 10 hottest days were rainy
hottest_10['rainfall'].value_counts(normalize=True)*100
```

```python
#finding 10 coldest days
coldest_10 = df.sort_values("maxtemp",ascending=True).head(10)
coldest_10
```

```python
#calculating how many of the 10 coldest days were rainy.
coldest_10['rainfall'].value_counts()
```

```python
#Calculating the percentage of 10 coldest days were rainy
coldest_10['rainfall'].value_counts(normalize=True)*100
```

```python
df.groupby("rainfall")['maxtemp'].mean()
```

```python
#Correlation analysis
df.corr(numeric_only=True)
```

```python
#Finding the weather conditions most associated with rain
df.groupby("rainfall")[["maxtemp","temparature","mintemp","humidity","cloud","sunshine","pressure","windspeed"]].mean()
```

```python
#difference between rainy and non-rainy averages
df.groupby("rainfall")["humidity"].mean().diff()
#findings: Rainy days have about 10.55% higher average humidity than non-rainy days
```

```python
df.groupby("rainfall")[["maxtemp","temparature","mintemp","humidity","cloud","sunshine","pressure","windspeed"]].mean().diff()
```

```python
#Overall conclusion:
#From the dataset, a typical rainy day tends to have:
# much more cloudy(+29.20), higher humidity(+10.55), somewhat higher windspeed(3.32), slightly lower temparature, less sunshine(-4.65), slightly lower pressure(-1.22)
```
